# Closing Bet Analysis

Analyze `data/closing_bet_results/*.jsonl` for gap distribution, exit scenarios, threshold sweeps, weekday effects, entry timing, liquidity bands, VI conditions, KOSDAQ regime, after-market adds, expectancy, and MDD.

```python
from pathlib import Path
import json
import pandas as pd

base = Path('../data/closing_bet_results')
rows = []
for path in sorted(base.glob('*.jsonl')):
    rows.extend(json.loads(line) for line in path.read_text(encoding='utf-8').splitlines() if line.strip())
df = pd.DataFrame(rows)
df.head()
```

```python
if not df.empty:
    display(df.groupby('exit_reason')['net_pnl'].agg(['count', 'mean', 'sum']))
    display(df.groupby('weekday')['net_pnl'].agg(['count', 'mean', 'sum']))
    for col, cuts in {
        'change_pct_at_entry': [0.05, 0.07, 0.10],
        'exec_strength_30m': [100, 115, 130],
    }.items():
        print(col)
        for cut in cuts:
            sub = df[df[col] >= cut]
            print(cut, len(sub), sub['net_pnl'].mean() if len(sub) else None)
```
